```python
import torch
import torch.nn as nn

# The formula for calculating output size is:
# Output_size = (Input_size - Kernel_size + 2 * Padding) / Stride + 1

print("--- Example 1: Basic Convolution (3x3 Kernel, Stride 1, Padding 0) ---")

# Create a sample input tensor: (Batch size=1, Channels=1, Height=10, Width=10)
input_tensor_1 = torch.randn(1, 1, 10, 10)

# Define a 2D convolution layer
# It takes 1 input channel and produces 1 output channel (feature map)
conv_layer_1 = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, stride=1, padding=0)

# Apply the convolution to the input tensor
output_tensor_1 = conv_layer_1(input_tensor_1)

print(f"Input tensor shape:  {input_tensor_1.shape}")
print(f"Output feature map shape: {output_tensor_1.shape}")
# Calculation: (10 - 3 + 2*0) / 1 + 1 = 8. The output is 8x8.
print("-> The 10x10 input has been reduced to 8x8.\n")

# ---

print("--- Example 2: Convolution with Padding (3x3 Kernel, Stride 1, Padding 1) ---")

input_tensor_2 = torch.randn(1, 1, 10, 10)

# Add padding=1. This adds a 1-pixel border around the image.
conv_layer_2 = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, stride=1, padding=1)
output_tensor_2 = conv_layer_2(input_tensor_2)

print(f"Input tensor shape:  {input_tensor_2.shape}")
print(f"Output feature map shape: {output_tensor_2.shape}")
# Calculation: (10 - 3 + 2*1) / 1 + 1 = 10. The output is 10x10.
print("-> Adding padding preserved the original 10x10 dimensions.\n")

# ---

print("--- Example 3: Convolution with Stride (3x3 Kernel, Stride 2, Padding 1) ---")

input_tensor_3 = torch.randn(1, 1, 10, 10)

# Change stride=2. The filter now moves 2 pixels at a time.
conv_layer_3 = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, stride=2, padding=1)
output_tensor_3 = conv_layer_3(input_tensor_3)

print(f"Input tensor shape:  {input_tensor_3.shape}")
print(f"Output feature map shape: {output_tensor_3.shape}")
# Calculation: (10 - 3 + 2*1) / 2 + 1 = 5. The output is 5x5.
print("-> Increasing the stride significantly reduced the output size to 5x5.")
```
