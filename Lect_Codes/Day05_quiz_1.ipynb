1. __init__(self): 모델의 "설계도"와 "부품 창고" 🏗️
__init__ 메서드는 모델 객체가 처음 생성될 때 단 한 번만 호출되는 생성자입니다. 이곳은 모델이라는 건물을 짓기 위해 필요한 모든 부품(재료)을 미리 선언하고 준비하는 공간입니다.

__init__의 역할
모델이 사용할 신경망 계층(Layers)들을 정의하고 초기화합니다.

비유하자면, 건물의 설계도를 그리고, 필요한 벽돌, 창문, 문과 같은 부품들을 창고에 준비해두는 단계입니다.

__init__에 준비할 수 있는 부품 목록
핵심 계산 계층 (Core Layers): nn.Linear, nn.Conv2d, nn.RNN, nn.LSTM 등 실질적인 계산을 수행하는 계층들.

활성화 함수 (Activation Functions): nn.ReLU, nn.Sigmoid, nn.Softmax 등 모델에 비선형성을 더해주는 함수들.

정규화 및 드롭아웃 (Normalization & Dropout): nn.BatchNorm2d, nn.Dropout 등 학습을 안정시키고 과대적합을 방지하는 계층들.

풀링 계층 (Pooling Layers): nn.MaxPool2d, nn.AvgPool2d 등 이미지 특징맵의 크기를 줄이는 계층들.

유틸리티 (Utilities): nn.Sequential 처럼 여러 계층을 하나의 블록으로 묶어주는 컨테이너.

2. forward(self, x): 모델의 "조립 설명서"와 "액션 플랜" ⚙️
forward 메서드는 모델 객체가 데이터(x)를 입력받아 호출될 때마다 매번 실제로 실행되는 부분입니다. __init__ 창고에 준비된 부품들을 가져와 어떤 순서로 조립하여 데이터의 흐름을 만들지 정의하는 공간입니다.

forward의 역할
입력 데이터가 모델의 첫 계층부터 마지막 계층까지 거치며 어떻게 계산되어 최종 예측값을 만들어내는지, 그 **'계산 흐름'**을 정의합니다.

비유하자면, 부품들을 순서대로 조립하는 방법을 상세히 적어놓은 조립 설명서와 같습니다.

forward에서 수행할 수 있는 작업 목록
__init__에서 선언한 계층 호출: x = self.layer1(x) 와 같이 준비된 부품을 사용합니다.

함수형 API 호출: F.relu(x) 처럼, 학습할 가중치가 없는 함수는 torch.nn.functional 라이브러리를 통해 바로 호출할 수 있습니다.

기본 텐서 연산: .view(), torch.cat(), +, * 등 모든 종류의 텐서 조작이 가능합니다.

파이썬 제어문 사용: if, for 문을 사용하여 입력에 따라 계산 흐름을 동적으로 바꿀 수 있습니다.

3. 계층 연결 및 설계: 실전 가이드 💡
모델을 설계하는 것은 __init__에서 부품을 준비하고 forward에서 조립하는 것입니다. 이때 가장 중요한 두 가지 원칙이 있습니다.

원칙 1: 파이프 연결하기 (입력과 출력 크기 맞추기)
가장 중요한 규칙입니다. 앞 계층의 출력 크기(out_features)는 반드시 다음 계층의 입력 크기(in_features)와 같아야 합니다.

```python
# layer1이 5개의 값을 출력하므로,
# layer2는 반드시 5개의 값을 입력으로 받아야 합니다.
self.layer1 = nn.Linear(in_features=10, out_features=5)
self.layer2 = nn.Linear(in_features=5, out_features=1)
```

원칙 2: 은닉층(Hidden Layer) 설계하기
은닉층의 구조는 모델의 성능을 결정하는 핵심 요소이며, 정답이 없는 예술의 영역입니다. 하지만 효과적인 몇 가지 기준이 있습니다.

깊이 (몇 개를 쌓을까?):

1~2개로 시작하세요. 대부분의 문제는 1~2개의 은닉층으로 충분합니다.

문제가 복잡할수록 더 깊게 쌓는 것을 고려할 수 있습니다.

너비 (뉴런을 몇 개로 할까?):

깔때기/피라미드 구조를 사용하세요. 입력층에서 시작해 점차 뉴런 수를 줄여나가는 방식(입력(64) -> 은닉1(32) -> 은닉2(16) -> 출력(1))이 가장 일반적입니다. 이는 모델이 정보의 핵심만 남기고 압축하도록 유도합니다.

보통 입력층과 출력층 뉴런 수 사이의 값으로 설정하며, **2의 거듭제곱(32, 64, 128...)**을 사용하는 것이 관례입니다.

활성화 함수:

nn.ReLU()를 기본으로 사용하세요. 특별한 이유가 없다면 은닉층의 활성화 함수는 ReLU가 가장 좋습니다.

활성화 함수는 모델에 비선형성을 부여하여, 선형 계층만으로는 풀 수 없는 복잡한 문제를 풀 수 있게 해줍니다.

결론적으로, 모델 설계는 __init__과 forward의 역할을 이해하고, "간단하고 합리적인 구조로 시작해서 실험을 통해 점진적으로 개선해나가는 과정"입니다.
