## 1. Dataset: "Recipe Book & Ingredient Warehouse of the Kitchen" 📜
The Dataset class defines all the information and access methods for the data itself that our model will learn.

- ***Key Roles***:
    - Knows what the entire data is, where it is located, and how many there are in total.
    - Defines how to retrieve the data from the storage according to the request "Get the nth data".

- ***Required Implementation Elements***:
    - __len__(self): Tells you how many total dishes (total size of the dataset) are prepared in our buffet.
    - __getitem__(self, index): If you say "Give me dish number 5", it accurately finds the data and answer (label) corresponding to dish number 5 and returns it as a single sample.

### 2. DataLoader: "Efficient Kitchen Staff & Serving Robot" 🤖
The DataLoader takes a recipe book called Dataset and takes charge of efficiently supplying (serving) data to the model (customer).

- ***Key Roles and Functions***:
    - Batch Composition (batch_size): Instead of bringing all the food from the buffet at once, it is provided by taking "10 pieces on a small plate (batch)". This is a key feature that allows efficient use of limited memory.
    - Data Shuffling (shuffle=True): By randomly shuffling the serving order each time, it prevents the model from being biased to a specific order of data.
    - Parallel Processing (num_workers): By having multiple kitchen assistants, it eliminates the waiting time in the learning process by preparing the next data in advance while the model is currently learning the data.

### 3. Why Use Them Together? Importance of "Separation of Roles"
The Dataset focuses only on the data itself, and the DataLoader focuses only on the technical aspects of efficiently supplying the data. By clearly separating roles in this way, there are the following advantages.

- ***Increased Code Clarity and Reusability***: The code becomes much cleaner and easier to understand. With the same Dataset (recipe book), you can flexibly combine and reuse it, such as using a DataLoader with shuffle=True when training and a DataLoader with shuffle=False when evaluating.

### 4. Additional Useful Features to Know
- TensorDataset: When there is data already prepared in tensor form, it is a **"meal kit"** that allows you to easily create a dataset with just one line of TensorDataset(X_tensor, y_tensor) without having to define class CustomDataset.

- collate_fn: It is a function that directly specifies **"Please bundle data that is difficult to bundle in a general way, such as text data with different lengths, into one batch in a special way"**.

- pin_memory=True: When learning with a GPU, it is an option like a **"highway bus exclusive lane"** that improves the data loading speed by preparing data in advance in a special memory space that allows data to be transferred faster.
