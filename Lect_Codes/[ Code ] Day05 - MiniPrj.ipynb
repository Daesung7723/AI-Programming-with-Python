### Mini-Project Proposal: Building Your Own Fashion Item Classifier (Fashion-MNIST)

#### 1. Project Goal
- You can load a new dataset (Fashion-MNIST) using torchvision.
- You can modify the input of an existing CNN model to suit the characteristics of the dataset (number of channels, image size).
- You can reuse the training and evaluation pipeline used for the CIFAR-10 classifier to train a new model and validate its performance.
- You can use the trained model to make predictions that classify real fashion item images.

#### 2. Dataset Introduction: Fashion-MNIST
- Fashion-MNIST is a dataset consisting of images of 10 different fashion items, including t-shirts, pants, sweaters, and dresses.
- Like CIFAR-10, it can be easily loaded using torchvision.datasets.
- Key Difference: CIFAR-10 images are color (3-channel) images sized 32x32, while Fashion-MNIST images are grayscale (1-channel) images sized 28x28. This difference is the key point of this project.
