**When training a deep learning model, we often need to change the shape of our data (tensors) to match the input format expected by the model. Freely manipulating a tensor's shape is like sculpting with clay—it's a core skill in data preprocessing.**

### 1. .view(-1, 1): From a 1D Vector to a 2D Matrix
***Key Role***  
.view(-1, 1) serves to reshape a tensor.  
- torch.arange(100): This creates a 1D tensor (vector) with 100 elements, like [0, 1, ..., 99]. Its shape is [100].
- .view(-1, 1): This converts the 1D tensor into a 2D tensor (matrix).
    - 1: "Fix the number of columns to be 1."
    - -1: "Since the number of columns is fixed at 1, automatically calculate the number of rows based on the total number of elements (100)."
    - Result: The shape changes from [100] to [100, 1].

***Visual Comparison:***
- Before ([100]): [0, 1, 2, ..., 99]
- After ([100, 1]): [[0], [1], [2], ..., [99]]

***Why is this transformation necessary?***
Most neural network layers in PyTorch, like nn.Linear, expect their input to be a 2D tensor with the shape (batch_size, num_features). torch.arange(100) can be seen as 100 data points, each with 1 feature. Therefore, we need to reshape it to (100, 1) so the model can correctly interpret it as "100 data points, each having 1 feature."

### 2. Essential Skills for Tensor Shaping
Besides .view(), there are several other useful functions you should know.

1) ***.view() vs .reshape(): The Safer Option ***  
While .view() and .reshape() perform almost the same function, .reshape() is smarter and safer. .view() only works if the tensor is stored contiguously in memory. .reshape() will create a copy in memory if needed to change the shape, making it less prone to errors.
**Conclusion:** Unless you have a specific reason, it's a better habit to use .reshape().

2) ***squeeze() & unsqueeze(): Squeezing and Expanding Dimensions***
These two functions are very useful for adding or removing dimensions of size 1.
- squeeze(): As the name implies, it "squeezes" out unnecessary dimensions of size 1.
    - Example: A tensor with shape [1, 10, 1] -> squeeze() -> shape [10]
- unsqueeze(dim): Adds a new dimension of size 1 at a specified position (dim).
  - Example: A tensor with shape [10] -> unsqueeze(0) -> shape [1, 10]

```python
a = torch.rand(1, 5, 1) # shape: [1, 5, 1]
b = a.squeeze()         # shape: [5]

c = torch.rand(5)       # shape: [5]
d = c.unsqueeze(0)      # shape: [1, 5]
e = c.unsqueeze(1)      # shape: [5, 1]
```

3) ***.permute(): Swapping Dimensions***
While .view() and .reshape() change the shape while keeping the order of elements, .permute() **swaps the order of the dimensions themselves.**
- Scenario: When dealing with image data, you might need to change the dimension order from (batch, channels, height, width) to (batch, height, width, channels).
- Example:

```python
# An image tensor with shape (32, 3, 28, 28)
image_tensor = torch.rand(32, 3, 28, 28)

# Change the order to (32, 28, 28, 3)
# This maps the old dimensions 0->0, 1->2, 2->3, 3->1
permuted_tensor = image_tensor.permute(0, 2, 3, 1)
```

If you can use these functions freely, you will be able to easily transform data of any shape into the format your model requires.
