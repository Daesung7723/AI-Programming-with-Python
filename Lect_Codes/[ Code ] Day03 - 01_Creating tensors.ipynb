```python
import torch
import numpy as np

# 1. Create from a Python list
py_list = [[1, 2, 3], [4, 5, 6]]
tensor_from_list = torch.tensor(py_list)
print("Tensor created from a Python list:")
print(tensor_from_list)
print("-" * 30)

# 2. Create from a NumPy array
# In this method, the NumPy array and the tensor share memory. (Changing one changes the other.)
numpy_array = np.array([[7, 8, 9], [10, 11, 12]])
tensor_from_numpy = torch.from_numpy(numpy_array)
print("Tensor created from a NumPy array:")
print(tensor_from_numpy)
print("-" * 30)

# 3. Create a tensor of a specific shape (very similar to NumPy)
shape = (2, 3) # 2 rows, 3 columns
zeros_tensor = torch.zeros(shape)
ones_tensor = torch.ones(shape)
rand_tensor = torch.rand(shape) # Uniform distribution between 0 and 1
randn_tensor = torch.randn(shape) # Normal distribution with mean 0 and standard deviation 1

print("All elements are 0 Tensor:")
print(zeros_tensor)
print("\nA tensor with all 1s:")
print(ones_tensor)
print("\nA tensor filled with random values:")
print(rand_tensor)
print("\nA tensor filled with values sampled from a normal distribution:")
print(randn_tensor)
```
