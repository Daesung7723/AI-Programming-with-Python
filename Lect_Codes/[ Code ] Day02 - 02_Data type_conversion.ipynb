## Converting Data Types in pandas
This guide will show you how to check and change the data type of columns in a DataFrame.

Step 1: Create a Sample DataFrame
Let's start with a DataFrame where the data types are not ideal. For example, Price is a string (object) instead of a number, and Stock is a decimal (float) instead of a whole number (int).

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with incorrect data types
data = {'ProductID': ['101', '102', '103', '104', '105'],
        'Price (LAK)': ['55,000', '120,000', '75,500', '98,000', '150,000'],
        'Stock': [15.0, 8.0, np.nan, 5.0, 11.0],
        'Last_Order_Date': ['2023-07-20', '2023-07-21', '2023-07-21', '2023-07-22', '2023-07-22']}
df = pd.DataFrame(data)

print("--- 1. Original DataFrame ---")
print(df)
```

Step 2: Check the Initial Data Types

Let's see what data types pandas assigned to our columns. We use df.info() for this. Notice ProductID and Price (LAK) are object (which means string).

```python
print("\n--- 2. Initial Data Types ---")
df.info()
```

Step 3: Convert Data Types

Now, let's fix the data types one by one.


3.1. Convert Price (LAK) from String to Number

The Price (LAK) column is a string because of the comma (,). We must remove it before we can convert it to a number.

```python
print("--- 3.1. Converting 'Price (LAK)' to a number ---")
# Create a copy to work on
df_converted = df.copy()

# Remove the comma and convert the column to a float (number with decimals)
df_converted['Price (LAK)'] = df_converted['Price (LAK)'].str.replace(',', '').astype(float)

print(df_converted)
print("\nNew Dtype for 'Price (LAK)':", df_converted['Price (LAK)'].dtype)
```

3.2. Convert Stock from Float to Integer

The Stock column should be an integer (a whole number) because you can't have half a product. Important: You cannot convert a column with NaN (missing values) to an integer. You must fill it first!

```python
print("\n--- 3.2. Converting 'Stock' to an integer ---")
# First, fill the missing value with 0
df_converted['Stock'].fillna(0, inplace=True)

# Now, convert the column to an integer
df_converted['Stock'] = df_converted['Stock'].astype(int)

print(df_converted)
print("\nNew Dtype for 'Stock':", df_converted['Stock'].dtype)
```

3.3. Convert Last_Order_Date from String to Datetime

To work with dates properly (e.g., calculate the time between dates), we should convert the string to a datetime object.

```python
print("\n--- 3.3. Converting 'Last_Order_Date' to datetime ---")
df_converted['Last_Order_Date'] = pd.to_datetime(df_converted['Last_Order_Date'])

print(df_converted)
print("\nNew Dtype for 'Last_Order_Date':", df_converted['Last_Order_Date'].dtype)
```

Step 4: Verify All Final Data Types

Let's use df.info() one last time to confirm that all our columns now have the correct data types.

```python
print("\n--- 4. Final Data Types ---")
# Also converting ProductID to an integer
df_converted['ProductID'] = df_converted['ProductID'].astype(int)
df_converted.info()
```
