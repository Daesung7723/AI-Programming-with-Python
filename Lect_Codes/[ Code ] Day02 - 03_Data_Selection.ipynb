## Selecting Data with .loc and .iloc
.loc selects data by label or name. Think L for Label.

.iloc selects data by integer position. Think I for Integer Index.

Step 1: Create a Sample DataFrame

First, let's create a DataFrame. We will set ProductID as our index (row label).

```python
import pandas as pd

# Create a sample DataFrame
data = {'ProductID': ['P101', 'P102', 'P103', 'P104', 'P105', 'P106'],
        'ProductName': ['Coffee', 'Tea', 'Juice', 'Water', 'Soda', 'Milk'],
        'Category': ['Drink', 'Drink', 'Drink', 'Drink', 'Drink', 'Dairy'],
        'Price_LAK': [15000, 12000, 18000, 5000, 8000, 25000],
        'Stock': [150, 200, 120, 300, 180, 90],
        'Origin_City': ['Pakse', 'Vientiane', 'Luang Prabang', 'Vientiane', 'Pakse', 'Savannakhet']}
df = pd.DataFrame(data)

# Set 'ProductID' as the index (row label)
df.set_index('ProductID', inplace=True)

print("--- 1. Original DataFrame ---")
print(df)
```

Step 2: Selecting Data with .loc (Label-Based)

Use .loc when you want to select data using the index labels (like P101, P103) and column names (like ProductName, Price_LAK).

2.1. Select a Single Row by Label

Let's get all the information for the product with ID P103.

```python
print("--- 2.1. Select single row 'P103' ---")
print(df.loc['P103'])
```

2.2. Select Multiple Rows by Labels

To select more than one row, give a list of labels.

```python
print("\n--- 2.2. Select rows 'P102' and 'P105' ---")
print(df.loc[['P102', 'P105']])
```

2.3. Select Rows and Specific Columns by Label

You can select a "slice" of your data. The format is df.loc[row_labels, column_labels].

```python
print("\n--- 2.3. Select rows 'P101' to 'P103' and columns 'ProductName' and 'Price_LAK' ---")
print(df.loc['P101':'P103', ['ProductName', 'Price_LAK']])
```

Step 3: Selecting Data with .iloc (Integer-Based)

Use .iloc when you want to select data using its position number (starting from 0), just like a Python list.

3.1. Select a Single Row by Position

Let's get the first row (position 0).

```python
print("--- 3.1. Select the first row (position 0) ---")
print(df.iloc[0])
```

3.2. Select the Last Row by Position

You can use negative numbers. -1 means the last row.

```python
print("\n--- 3.2. Select the last row (position -1) ---")
print(df.iloc[-1])
```

3.3. Select Rows and Columns by Position

The format is df.iloc[row_positions, column_positions]. Let's get the first 3 rows and the first 2 columns.

```python
print("\n--- 3.3. Select first 3 rows (0,1,2) and first 2 columns (0,1) ---")
print(df.iloc[0:3, 0:2])
```

Step 4: Conditional Selection (Very Powerful!)

This is a very common task. You can use .loc to select rows that meet a certain condition.

4.1. Select Rows Based on a Condition

Let's find all products that are from Vientiane.

```python
print("--- 4.1. Select all products from 'Vientiane' ---")
print(df.loc[df['Origin_City'] == 'Vientiane'])
```
