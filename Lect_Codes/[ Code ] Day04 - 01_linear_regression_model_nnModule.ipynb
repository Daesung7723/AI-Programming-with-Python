```python
import torch
import torch.nn as nn

# nn.Module must be accepted as an exception.
class LinearRegressionModel(nn.Module):
    def __init__(self):
        """Calls the external module used by the model."""
        super().__init__()
        # Centered on one input (in_features) and one right-hand side (out_features)
        # Define a crossover (linear layer).
        self.linear = nn.Linear(in_features=1, out_features=1, bias=True)

    def forward(self, x):
        """Defines the order in which data is transmitted (forward propagation)."""
        return self.linear(x)

# -------------------------------------------------------------

#Create with photos
model = LinearRegressionModel()
print(model)
print("-" * 50)


# Enter the format to generate user (dummy) data.
# Input size: 10 samples, each sample has 1 left -> torch.Size([10, 1])
input_tensor = torch.randn(10, 1)

# Pass data to the model to get the passed values (run forward propagation)
output_tensor = model(input_tensor)

print(f"Input shape: {input_tensor.shape}")
print(f"Output shape: {output_tensor.shape}")
print("-" * 50)


# The model outputs a large number of people (weights and biases).
# Accessible through the .weight and .bias properties.
print(f"linear.weight {model.linear.weight.data} # Randomly initialized weights (w)")
print(f"linear.bias   {model.linear.bias.data} # Randomly initialized bias (b)")
```
