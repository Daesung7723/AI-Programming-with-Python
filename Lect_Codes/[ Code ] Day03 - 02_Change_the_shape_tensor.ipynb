```python
# Tensor of shape (3, 4)
t = torch.rand(3, 4)
print(f"Original shape: {t.shape}") # (3, 4)

# 1. view
t_view = t.view(2, 6)
print(f"\nview(2, 6) shape: {t_view.shape}") # (2, 6)

# 2. unsqueeze: Add dimension
# Add dimension at position dim=0 -> (1, 3, 4)
# Often used when adding a dimension to a batch of image data
t_unsqueezed = t.unsqueeze(dim=0)
print(f"unsqueeze(0) shape: {t_unsqueezed.shape}")

# Add dimension at position dim=2 -> (3, 4, 1)
# Often used when adding a dimension to the channel of a grayscale image used
t_unsqueezed_2 = t.unsqueeze(dim=2)
print(f"unsqueeze(2) shape: {t_unsqueezed_2.shape}")

# 3. squeeze: remove dimension
# (1, 3, 4) -> (3, 4)
t_squeezed = t_unsqueezed.squeeze(dim=0)
print(f"\nsqueeze(0) shape: {t_squeezed.shape}")
```
