```python
import torch

# --- Step 1: Prepare the Data ---
# We will create sample data based on the true relationship y = 2x + 1
# Our goal is for the model to learn W=2 and b=1.
x_train = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y_train = torch.tensor([[3.0], [5.0], [7.0], [9.0]]) # y = 2x + 1

# --- Step 2: Initialize Model Parameters ---
# Start with random values for Weight (W) and Bias (b).
# requires_grad=True tells PyTorch to calculate gradients for these tensors.
W = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

print(f"Initial W: {W.item():.4f}, Initial b: {b.item():.4f}")
print("-" * 30)

# --- Step 3: Set Hyperparameters ---
# Hyperparameters are settings that we choose to control the training process.
learning_rate = 0.01
epochs = 2000 # The number of times we will go through the training data

# --- Step 4: Training Loop (Gradient Descent) ---
for epoch in range(epochs):
    # 1. Forward Pass: Calculate the model's prediction (Hypothesis)
    hypothesis = x_train * W + b

    # 2. Calculate Loss: How wrong is the model?
    # We use Mean Squared Error (MSE) as our loss function.
    loss = torch.mean((hypothesis - y_train) ** 2)

    # 3. Backward Pass: Calculate gradients
    # Before calculating new gradients, we must clear the old ones.
    if W.grad is not None:
        W.grad.zero_()
    if b.grad is not None:
        b.grad.zero_()
    
    # loss.backward() calculates the gradient of the loss with respect to
    # every tensor that has requires_grad=True.
    loss.backward()

    # 4. Update Parameters: Move W and b in the direction that reduces the loss.
    # We manually perform the gradient descent step.
    # We use torch.no_grad() to prevent this update step from being tracked
    # in the computation graph.
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

    # --- Optional: Print progress ---
    if (epoch + 1) % 100 == 0:
        print(f'Epoch {epoch+1:4d}/{epochs} | Loss: {loss.item():.4f} | W: {W.item():.4f} | b: {b.item():.4f}')

# --- Step 5: Check the Final Result ---
print("-" * 30)
print(f"Training complete!")
print(f"Final W (should be close to 2): {W.item():.4f}")
print(f"Final b (should be close to 1): {b.item():.4f}")
```
