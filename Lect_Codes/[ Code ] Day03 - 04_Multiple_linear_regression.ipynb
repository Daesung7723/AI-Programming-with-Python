```python
import torch

# --- Step 1: Prepare the Data ---
# We will create data with two input features (x1, x2).
# The true relationship is y = 2*x1 + 3*x2 + 0.5
# Our goal is for the model to learn W=[2, 3] and b=0.5.
x_train = torch.tensor([[1.0, 1.0],
                        [2.0, 3.0],
                        [3.0, 2.0],
                        [4.0, 5.0]])

y_train = torch.tensor([[5.5],   # 2*1 + 3*1 + 0.5
                        [13.5],  # 2*2 + 3*3 + 0.5
                        [12.5],  # 2*3 + 3*2 + 0.5
                        [23.5]]) # 2*4 + 3*5 + 0.5

# --- Step 2: Initialize Model Parameters ---
# W is now a 2x1 vector to match the two input features.
# b remains a single scalar value.
W = torch.randn((2, 1), requires_grad=True) # Shape: (num_features, 1)
b = torch.randn(1, requires_grad=True)

print(f"Initial W: {W.squeeze().tolist()}, Initial b: {b.item():.4f}")
print("-" * 40)

# --- Step 3: Set Hyperparameters ---
learning_rate = 0.01
epochs = 2000

# --- Step 4: Training Loop (Gradient Descent) ---
for epoch in range(epochs):
    # 1. Forward Pass: Calculate the model's prediction (Hypothesis)
    # Use matrix multiplication (matmul) for multiple features.
    # (4, 2) @ (2, 1) -> (4, 1) which matches y_train's shape.
    hypothesis = torch.matmul(x_train, W) + b

    # 2. Calculate Loss (MSE)
    loss = torch.mean((hypothesis - y_train) ** 2)

    # 3. Backward Pass: Calculate gradients
    # Clear previous gradients before calculating new ones.
    if W.grad is not None:
        W.grad.zero_()
    if b.grad is not None:
        b.grad.zero_()
    
    loss.backward()

    # 4. Update Parameters
    # Use torch.no_grad() to prevent tracking this operation.
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

    # --- Optional: Print progress ---
    if (epoch + 1) % 100 == 0:
        # .squeeze() removes extra dimensions for cleaner printing
        w_list = W.squeeze().tolist()
        print(f'Epoch {epoch+1:4d}/{epochs} | Loss: {loss.item():.4f} | W: [{w_list[0]:.4f}, {w_list[1]:.4f}] | b: {b.item():.4f}')

# --- Step 5: Check the Final Result ---
print("-" * 40)
print(f"Training complete!")
final_w_list = W.squeeze().tolist()
print(f"Final W (should be close to [2.0, 3.0]): [{final_w_list[0]:.4f}, {final_w_list[1]:.4f}]")
print(f"Final b (should be close to 0.5): {b.item():.4f}")

```
