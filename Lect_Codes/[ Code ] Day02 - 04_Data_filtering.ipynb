## Filtering Data in pandas
Step 1: Create a Sample DataFrame

Let's start with a dataset of products. We will use this DataFrame to practice filtering.

```python
import pandas as pd

# Create a sample DataFrame
data = {'ProductID': ['P101', 'P102', 'P103', 'P104', 'P105', 'P106', 'P107'],
        'ProductName': ['Lao Coffee', 'Green Tea', 'Sticky Rice', 'Water', 'Lao Beer', 'Milk', 'Spicy Peanuts'],
        'Category': ['Drink', 'Drink', 'Food', 'Drink', 'Drink', 'Dairy', 'Snack'],
        'Price_LAK': [18000, 12000, 25000, 5000, 10000, 22000, 15000],
        'Stock': [150, 200, 300, 500, 250, 90, 180],
        'Origin_City': ['Pakse', 'Vientiane', 'Vientiane', 'Luang Prabang', 'Vientiane', 'Savannakhet', 'Pakse']}
df = pd.DataFrame(data)

print("--- 1. Original DataFrame ---")
print(df)
```

Step 2: Filtering with a Single Condition

This is the most basic type of filtering.

2.1. Filter by Text Value

Let's find all the products that are in the 'Food' category.

```python
print("--- 2.1. Filter for all products in the 'Food' Category ---")
# The condition is df['Category'] == 'Food'
food_products = df[df['Category'] == 'Food']
print(food_products)
```

2.2. Filter by a Number

Now, let's find all products that cost more than 20,000 LAK.

```python
print("\n--- 2.2. Filter for products with Price_LAK > 20000 ---")
expensive_products = df[df['Price_LAK'] > 20000]
print(expensive_products)
```

Step 3: Filtering with Multiple Conditions

You can combine conditions using & (AND) and | (OR). Remember to wrap each condition in parentheses ().

3.1. Using the & (AND) Operator

Let's find all 'Drink' products that are from 'Vientiane'. Both conditions must be true.

```python
print("--- 3.1. Filter for 'Drink' products AND from 'Vientiane' ---")
condition = (df['Category'] == 'Drink') & (df['Origin_City'] == 'Vientiane')
print(df[condition])
```

3.2. Using the | (OR) Operator

Let's find products that are either from 'Pakse' OR have a stock level less than 100. Only one of the conditions needs to be true.

```python
print("\n--- 3.2. Filter for products from 'Pakse' OR Stock < 100 ---")
condition = (df['Origin_City'] == 'Pakse') | (df['Stock'] < 100)
print(df[condition])
```

Step 4: Filtering with .isin()

When you want to match several possible values in a column, using .isin() is cleaner than writing many OR conditions.

4.1. Find Products from Specific Cities

Let's find all products from 'Pakse' and 'Luang Prabang'.

```python
print("--- 4.1. Filter for products from 'Pakse' or 'Luang Prabang' using .isin() ---")
cities_to_find = ['Pakse', 'Luang Prabang']
print(df[df['Origin_City'].isin(cities_to_find)])
```
