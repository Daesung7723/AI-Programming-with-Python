## Grouping and Aggregating Data with groupby()
Step 1: Create a Sample DataFrame

Let's create our product dataset. It has categorical columns we can group by (Origin_City, Category) and numerical columns we can summarize (Price_LAK, Stock).

```python
import pandas as pd

# Create a sample DataFrame
data = {'Origin_City': ['Vientiane', 'Pakse', 'Vientiane', 'Luang Prabang', 'Pakse', 'Vientiane', 'Luang Prabang'],
        'Category': ['Drink', 'Snack', 'Drink', 'Drink', 'Snack', 'Food', 'Food'],
        'ProductName': ['Lao Tea', 'Spicy Peanuts', 'Lao Beer', 'Water', 'Crispy Rice', 'Sticky Rice', 'Khao Soi'],
        'Price_LAK': [12000, 15000, 10000, 5000, 13000, 25000, 30000],
        'Stock': [200, 180, 250, 500, 150, 300, 80]}
df = pd.DataFrame(data)

print("--- 1. Original DataFrame ---")
print(df)
```

Step 2: Simple Grouping with One Aggregation

Let's group our data by Origin_City and find the average price and stock for products from each city.

(Note: numeric_only=True is used to tell mean() to only calculate for number columns.)

```python
print("--- 2. Group by 'Origin_City' and find the mean ---")
city_summary = df.groupby('Origin_City').mean(numeric_only=True)
print(city_summary)
```


You can easily change the aggregation function. For example, let's find the total stock (sum()) in each city.

```python
print("\n--- Find the total stock for each city using sum() ---")
city_stock = df.groupby('Origin_City')['Stock'].sum()
print(city_stock)
```

Step 3: Grouping by Multiple Columns

You can group by more than one column to get a more detailed summary. Let's group by both Origin_City and Category.

```python
print("--- 3. Group by 'Origin_City' AND 'Category' and find the mean ---")
detailed_summary = df.groupby(['Origin_City', 'Category']).mean(numeric_only=True)
print(detailed_summary)
```

Step 4: Using .agg() for Multiple Aggregations

The .agg() method is the most powerful tool. It lets you perform several different aggregations at the same time.

Let's group by Category and for each category, find:
- The sum and average (mean) of the Stock.
- The minimum and maximum Price_LAK.

```python
print("--- 4. Using .agg() for multiple custom aggregations ---")
aggregation_rules = {
    'Stock': ['sum', 'mean'],
    'Price_LAK': ['min', 'max']
}

category_summary = df.groupby('Category').agg(aggregation_rules)
print(category_summary)
```

Step 5: Grouping without Setting an Index

Sometimes, you want to keep your grouped columns as regular columns instead of an index. You can do this with as_index=False.

```python
print("--- 5. Grouping with as_index=False ---")
summary_as_df = df.groupby('Origin_City', as_index=False).mean(numeric_only=True)
print(summary_as_df)
```
