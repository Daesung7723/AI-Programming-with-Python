```python
import torch

# ===================================================================
# 1. Setup: Prepare Data and Model Parameters
# ===================================================================
# Create simple data with the relationship y = 2x
x = torch.tensor(1.0)
y = torch.tensor(2.0)

# Initialize the weight (w) to be trained with an arbitrary value
# Set requires_grad=True to track gradients for this tensor
w = torch.tensor(1.0, requires_grad=True)

# ===================================================================
# 2. Forward Propagation: Calculate Prediction and Loss
# ===================================================================
# Calculate the prediction (y_pred) using the current weight w
y_pred = w * x

# Calculate the loss using MSE (Mean Squared Error)
loss = (y_pred - y)**2

print(f"Input (x): {x.item()}, True Answer (y): {y.item()}")
print(f"Initial Weight (w): {w.item():.2f}")
print(f"Current Prediction (y_pred): {y_pred.item():.2f}")
print(f"Current Loss: {loss.item():.2f}")
print("-" * 40)

# ===================================================================
# 3. Backpropagation: Calculate Gradients
# ===================================================================
# -------------------------------------------------------------------
# 3-1. Manually Calculate Gradient (Differentiation)
# -------------------------------------------------------------------
# Our goal is to find the value of "d(loss)/d(w)".
# Since loss = (w * x - y)^2, differentiating with the Chain Rule gives:
# d(loss)/d(w) = 2 * (w * x - y) * x.
manual_grad = 2 * (w * x - y) * x
print(f"Manually Calculated Gradient d(loss)/d(w): {manual_grad.item():.2f}")

# -------------------------------------------------------------------
# 3-2. Automatically Calculate Gradient using PyTorch's backward()
# -------------------------------------------------------------------
# Calling backward() on the loss tensor
# calculates the gradients for all tensors with requires_grad=True that affected the loss.
loss.backward()
print(f"PyTorch Auto-Calculated Gradient (w.grad): {w.grad.item():.2f}")
print("-" * 40)

# ===================================================================
# 4. Weight Update (Gradient Descent)
# ===================================================================
# Update the weight w using the calculated gradient (w.grad).
# Slightly modify w in the opposite direction of the gradient to reduce the loss.
learning_rate = 0.1
new_w = w - learning_rate * w.grad

print(f"Learning Rate: {learning_rate}")
print(f"Weight before update (w): {w.item():.2f}")
print(f"Weight after update (new_w): {new_w.item():.2f}")
print("\n-> w has moved slightly closer to the correct answer, 2.0.")
print("Repeating this process is exactly how a deep learning model learns.")
```
