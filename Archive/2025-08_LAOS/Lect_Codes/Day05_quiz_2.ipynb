### 1. The Core Role of Optimizers
The core role of an optimizer is to look at the map of 'gradients' calculated through loss.backward() and actually modify (update) the model's trainable parameters (weights and biases) according to that map.
In other words, it is the executive who moves the model's parameters step by step in the direction of reducing the loss, and the ***engine*** that practically advances learning.

- ***Position in the Learning Loop***
The optimizer is executed with the command optimizer.step() in the last step of the three steps of the learning loop.
    1. optimizer.zero_grad(): (Clear gradients)
    2. loss.backward(): (Calculate map - Backpropagation)
    3. optimizer.step(): (Update actual parameters with the calculated map - Role of the optimizer!)

### 2. Loss Function vs. Optimizer: Clearly Distinguishing Roles
It is easy to confuse the roles of these two, but they are clearly different.

- Loss Function: Determines "what" to optimize.
    - Role: Defines how to calculate the 'error', i.e., how wrong the model's predictions are. In other words, it determines the ***shape of the mountain*** we have to descend.
    - Varies depending on the problem type:
        - Regression: nn.MSELoss (distance between predicted and actual values)
        - Classification: nn.CrossEntropyLoss (difference between predicted probability distribution and actual distribution)
    - Optimizer: Determines "how" to optimize.
        - Role: Based on the 'error' calculated by the loss function, it defines ***the 'method' of which strategy to update the parameters***. In other words, it determines 'how to descend' the determined 'mountain'.
        - Irrelevant to the problem type: Mountaineering techniques such as Adam and SGD are not limited by the type of mountain (regression, classification). They can be used universally for any problem.

***Conclusion: The loss function determines "what kind of error to reduce?", and the optimizer determines "in what way to reduce that error?".***

### 3. Types of Main Optimizers
Each optimizer is like a mountaineer with its own strategy.

##### 1) SGD (Stochastic Gradient Descent)
- Analogy: The most basic mountaineer 🚶‍♂️
- Explanation: It is the simplest and most honest method of going down just one step in the direction with the steepest slope from the current position.
- Advantages: The concept is simple and the calculation is light.
- Disadvantages: The learning speed is slow, or it can be trapped in a narrow valley (local minimum).

##### 2) Momentum
- Analogy: A ball with inertia (acceleration) 🏀
- Explanation: It is an improved version of SGD that remembers the direction and speed (inertia) it was descending before and reflects it in the next step.
- Advantages: Thanks to inertia, it can ignore and pass through small valleys, and the learning speed is faster than SGD.

##### 3) Adam (Adaptive Moment Estimation)
- Analogy: A professional mountaineer equipped with all the latest equipment 🧗‍♀️
- Explanation: It is the most widely used, all-purpose optimizer with good performance to the point that it is said to **"just use Adam if you don't know what to use."** It cleverly combines Momentum and RMSprop methods.
- Advantages: It automatically adjusts the stride (learning rate) for each parameter and finds the optimal point quickly and stably.

##### 4) RMSprop (Root Mean Square Propagation)
- Analogy: A mountaineer who adjusts the stride according to the terrain 👟
- Explanation: It adaptively adjusts the learning rate by reducing the stride for parameters with many recent changes and increasing the stride for parameters with few changes.
- Advantages: It stabilizes learning by applying different learning rates for each parameter.
