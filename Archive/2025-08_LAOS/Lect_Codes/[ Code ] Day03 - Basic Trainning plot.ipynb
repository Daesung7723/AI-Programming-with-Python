```python
import torch
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider
from IPython.display import display

# ===================================================================
# 1. Training and Visualization Function
# ===================================================================
# The function now accepts both epochs and learning_rate as arguments.
def train_and_plot(epochs, learning_rate):
    # --- Setup: Prepare Data and Model Parameters ---
    x = torch.tensor(1.0)
    y = torch.tensor(2.0)
    w = torch.tensor(1.0, requires_grad=True) # Reset w for each run

    # Lists to store the history for plotting
    w_history = []
    loss_history = []

    # --- Training Loop ---
    for epoch in range(epochs):
        y_pred = w * x
        loss = (y_pred - y)**2

        w_history.append(w.item())
        loss_history.append(loss.item())

        if w.grad is not None:
            w.grad.zero_()
        
        loss.backward()

        with torch.no_grad():
            w -= learning_rate * w.grad

    # --- Visualize the Learning Process ---
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Graph 1: Loss over Epochs
    axes[0].plot(loss_history)
    axes[0].set_title(f'Loss Convergence (Epochs={epochs}, LR={learning_rate})')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].grid(True)

    # Graph 2: Weight (w) over Epochs
    axes[1].plot(w_history)
    axes[1].axhline(y=2.0, color='r', linestyle='--', label='True Value (2.0)')
    axes[1].set_title(f'Weight (w) Convergence (Epochs={epochs}, LR={learning_rate})')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Weight (w)')
    axes[1].legend()
    axes[1].grid(True)

    plt.suptitle('Visualization of the Learning Process', fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# ===================================================================
# 2. Create and Display the Interactive Sliders
# ===================================================================
# Create an integer slider for the number of epochs
epoch_slider = IntSlider(
    min=10, 
    max=500, 
    step=10, 
    value=100, 
    description='Epochs:',
    continuous_update=False
)

# Create a float slider for the learning rate
learning_rate_slider = FloatSlider(
    min=0.001,
    max=0.1,
    step=0.001,
    value=0.01,
    description='Learning Rate:',
    continuous_update=False,
    readout_format='.3f' # Display value with 3 decimal places
)

# Link both sliders to the function using `interactive`
interactive_plot = interactive(train_and_plot, epochs=epoch_slider, learning_rate=learning_rate_slider)

# Display the sliders and the output area for the graphs
print("Use the sliders to change epochs and learning rate to see the results.")
display(interactive_plot)
```
