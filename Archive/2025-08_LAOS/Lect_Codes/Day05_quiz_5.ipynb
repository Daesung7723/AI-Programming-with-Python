**Creating a deep learning model in PyTorch is like creatively assembling 'Lego blocks' called various types of **Layers**. Each layer performs a unique role specialized in processing specific types of data and extracting features.**

#### 1. nn.Linear: The Most Basic and Versatile Block
nn.Linear is called a **Fully Connected Layer** and is the most basic and universally used layer.

- ***Code***: 
    - nn.Linear(in_features=10, out_features=5)
- ***Meaning***: 
    - This layer takes 10 input features and transforms them into 5 output features. Internally, it performs a linear transformation (y = Wx + b) by multiplying the input data by a weight matrix (W) and adding a bias (b).
- ***Input and Output Size***:
    - Input: A tensor where each data sample consists of 10 numbers. (Batch size, 10) shape.
    - Output: 10 pieces of input information are summarized or transformed into 5 new pieces of information. (Batch size, 5) shape.
- ***Main Uses***:
    - When learning features of structured data such as tables or Excel.
    - When finally combining features that have passed through other types of layers (e.g., CNN, RNN) to derive a conclusion.

#### 2. Special Layers According to Data Type
nn.Linear alone cannot effectively process all types of data. Therefore, special layers suitable for the characteristics of the data must be used.

##### 1) For Processing Image Data (Photos, Pictures, etc.)
- ***nn.Conv2d (Convolutional Layer)***  
    - ***Role***: Extracts visual features such as lines, edges, and textures by scanning over the image like a 'magnifying glass'. It is specialized in learning while maintaining the spatial information of the image. (Core content of Day 5)
- ***nn.MaxPool2d (Max Pooling Layer)***  
    - ***Role***: Reduces the size by leaving only the most important features (the largest value) among the features found by the convolutional layer. This increases computational efficiency and makes the model less sensitive (robust) to minute changes in the position of objects.

##### 2) For Processing Sequential Data (Text, Voice, Time Series Stock Prices, etc.)
- ***nn.RNN (Recurrent Neural Network Layer)***  
    - ***Role***: A layer with **'memory'**. It remembers the calculation results of the previous step (e.g., the previous word) and reflects it in the calculation of the current step (the current word). It is used to process data where the order of words or the flow of time is important. (Core content of Day 6)
- ***nn.LSTM (Long Short-Term Memory Layer)***  
    - ***Role***: An advanced form of RNN, a layer with enhanced 'long-term memory' ability that can effectively remember information from the more distant past. It is mainly used when dealing with complex sentences or long time-series data.

#### 3. Layers for Learning Assistance
- ***nn.Dropout (Dropout)***  
    - ***Role***: A regularization device that prevents overfitting, where the model becomes too dependent on specific neurons, by intentionally making some neurons rest randomly during the learning process.
