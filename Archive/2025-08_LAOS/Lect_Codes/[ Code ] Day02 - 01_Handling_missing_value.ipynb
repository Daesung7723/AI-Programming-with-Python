## Handling Missing Values (NaN) in pandas
This guide shows you how to find, remove, and replace missing data in a pandas DataFrame.

Step 1: Create a Sample DataFrame
First, let's create a DataFrame with some missing values (np.nan). This will be our sample data for the exercises.

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values
data = {'Name': ['John', 'Jane', 'Peter', 'Sue', 'Tom', np.nan],
        'Age': [24, np.nan, 31, 29, np.nan, 25],
        'Score': [88, 92, np.nan, 75, 95, 68],
        'City': ['Vientiane', 'Pakse', 'Vientiane', np.nan, 'Luang Prabang', 'Pakse']}
df = pd.DataFrame(data)

print("--- 1. Original DataFrame ---")
print(df)
```

Step 2: Find Missing Values
Before we can fix missing values, we need to find them.

2.1. Check with isna()
The isna() method returns a DataFrame of the same size, but with True for missing values and False for non-missing values.

```python
print("--- 2.1. Check where values are missing (True = Missing) ---")
print(df.isna())
```

2.2. Count Missing Values with sum()

This is a very useful command. It tells you exactly how many missing values are in each column.

```python
print("\n--- 2.2. Count of missing values in each column ---")
print(df.isna().sum())
```

Step 3: Remove Missing Values (dropna)
One way to handle missing data is to simply remove it.

3.1. Drop Rows with Missing Values
This removes any row that contains at least one NaN. Notice that the rows for 'Jane' (missing Age) and 'Tom' (missing Age) are gone.

```python
print("--- 3.1. Drop all rows that contain a missing value ---")
df_dropped_rows = df.dropna()
print(df_dropped_rows)
```

3.2. Drop Columns with Missing Values
This removes any column that contains at least one NaN. Since every column except 'Name' has a missing value, they are all removed.

```python
print("\n--- 3.2. Drop all columns that contain a missing value ---")
df_dropped_cols = df.dropna(axis='columns')
print(df_dropped_cols)
```

Step 4: Replace (Fill) Missing Values (fillna)
Removing data isn't always the best option. Sometimes, it's better to replace the missing values. This is also called imputation.

4.1. Fill with a Specific Value (e.g., 0)
You can replace all NaN values with a single number, like 0.

```python
print("--- 4.1. Fill all missing values with 0 ---")
df_filled_zero = df.fillna(0)
print(df_filled_zero)
```

4.2. Fill with the Column's Mean (Average)

A smarter way is to fill missing numbers with the average of that column. Let's do this for the 'Age' column.

```python
print("\n--- 4.2. Fill missing 'Age' with the average age ---")
# First, calculate the mean of the 'Age' column
mean_age = df['Age'].mean()
print(f"The average age is: {mean_age:.2f}\n")

# Create a copy to keep the original df clean
df_filled_mean = df.copy()

# Now, fill the missing values in the 'Age' column
df_filled_mean['Age'].fillna(mean_age, inplace=True)
print(df_filled_mean)
```

4.3. Fill with the Previous Value (Forward Fill)

For categorical data like 'City', you can fill NaN with the value from the row above it. This is called a forward fill (ffill).

```python
print("\n--- 4.3. Fill missing 'City' with the value from the row above ---")
df_filled_ffill = df.copy()
df_filled_ffill['City'].fillna(method='ffill', inplace=True)
print(df_filled_ffill)
```
