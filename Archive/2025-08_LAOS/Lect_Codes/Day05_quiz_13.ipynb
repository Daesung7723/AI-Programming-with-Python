***The first hyperparameter to suspect when training is unstable is the learning rate.***

### 1. When Loss Explodes or Becomes NaN
- **Cause:** The learning rate is too high. The updates to the model's parameters are too large, causing the optimizer to overshoot the minimum of the loss function. This results in the parameters landing in a region with even higher loss, and this effect compounds with each step, leading to an exponential increase in the loss value or a numerical overflow (NaN).
- **Solution:** Decrease the learning rate. A common practice is to reduce it by a factor of 10 (e.g., from 0.1 to 0.01, or 0.01 to 0.001).

### 2. When Loss Does Not Decrease (Stagnates)
- **Cause:** The learning rate is too low. The updates to the parameters are so small that the model makes almost no progress in descending the loss landscape. This leads to extremely slow training, or the model may get stuck in a shallow local minimum without enough momentum to escape.
- **Solution:** Increase the learning rate. A common practice is to increase it by a factor of 10 (e.g., from 0.0001 to 0.001).
