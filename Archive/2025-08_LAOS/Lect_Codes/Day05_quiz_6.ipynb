**When training a deep learning model, we need to efficiently process a vast amount of data. It is very important to clearly understand the concepts and relationships of Epoch, Batch Size, and Iteration, which are the most basic training units used at this time.**

    These concepts can be explained by analogy to **"the process of studying with a thick textbook."**

#### 1. Definition of Terms
##### 1) Epoch 📖
- ***Definition***: It refers to one training cycle in which the entire dataset is used once from beginning to end to complete learning.
- ***Analogy***: It is like **"reading a math textbook once from page 1 to the last page."** If num_epochs = 10, it means reading the textbook a total of 10 times repeatedly.

##### 2) Batch Size 📄
- ***Definition***: It refers to the number of data included in one bundle when dividing the entire dataset into several small bundles (batches). The model updates the parameters once after learning this batch.
- ***Analogy***: Since it is difficult to read a thick textbook all at once, it is like **"reading by cutting off 10 pages at a time."** Here, batch_size becomes 10.

##### 3) Iteration 🚶‍♂️
- ***Definition***: It refers to the number of batches required to complete 1 epoch, that is, the number of parameter updates. Iteration is not a value that the user directly sets, but is automatically determined by the total number of data and the batch size.
- ***Analogy***: If you read a 1,000-page textbook by cutting it into 10 pages, you will have to read it a total of 100 times to read one book. Here, 100 'reading actions' is iteration.
- ***Calculation***: Iteration = Total number of data / Batch size

#### Example: If there are 2,000 data and batch_size is 100,
- 20 iterations (2000 / 100 = 20) are required to complete 1 epoch.
- In other words, if you repeat the process of looking at 100 data and updating the parameters 20 times, you will see the entire dataset once and 1 epoch will be completed.

#### 2. Criteria for Determining Batch Size
 The batch size is a key hyperparameter that directly affects the learning speed and final performance of the model. There is no right answer, but there are general criteria and trade-off relationships.

##### Practical Approach
1. ***Make it as large as GPU memory allows***: The most realistic first constraint to determine the batch size is GPU memory. If a "CUDA out of memory" error occurs, it means that the batch size is too large, so you should reduce it immediately.
2. ***Use powers of 2 (convention)***: It is common to use powers of 2 such as 32, 64, 128, 256... This is because it maximizes the efficiency of the hardware and increases the data processing speed. It is best to start with 32 or 64 unless there is a special reason.

##### Advantages and disadvantages of batch size
Features|Large batch size (e.g., 256) | Small batch size (e.g., 16)
---|---|---
Learning speed|Fast (Maximize GPU parallel operation efficiency)|Slow (Data must be read frequently)
Memory usage|Many|Few
Learning stability|Stable (Reflects the tendency of the entire data well)|Unstable (Can be swayed by noise in a small number of data)
Final performance|Can reach a stable optimum (Sharp Minimum)|Thanks to instability, you may find a better optimum (Flat Minimum) (Generalization performance may be better)
