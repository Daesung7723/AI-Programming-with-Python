#### It's good to remember the 3 steps of the learning loop as a flow of "Clean - Calculate - Apply".

***1. optimizer.zero_grad() (Step 1: Clean)***
- ***Role***: Cleanly initializes the gradient values that were calculated in the learning of the previous batch.
- ***Reason***: PyTorch does not overwrite the new gradient each time loss.backward() is called, but operates by adding (accumulating) to the existing gradient. Therefore, before accurately calculating the gradient of each new batch, the old gradient information of the previous batch must be cleared.

***2. loss.backward() (Step 2: Calculate)***
- ***Role***: Automatically performs backpropagation from the loss value, calculating the gradient that indicates how much each trainable parameter of the model affected the loss.
- ***Analogy***: This is the step of calculating the direction and slope of "which direction is the steepest downhill path to descend the mountain".

***3. optimizer.step() (Step 3: Apply)***
- ***Role***: Based on the gradient values calculated in loss.backward(), the optimizer (e.g., Adam, SGD) actually modifies the model's parameters (weights and biases) according to its update rules.
