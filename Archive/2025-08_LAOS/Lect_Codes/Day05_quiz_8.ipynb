The loss function measures **how different (error)** the model's prediction is from the actual answer. Since the model's parameters are updated in the direction of minimizing this error value, the loss function plays a very important role in determining the learning goal and direction of the model.

The appropriate loss function depends on the type of problem, and they are largely divided into those for **Regression** and **Classification**.

#### 1. Regression Loss Functions
Regression is a problem that predicts continuous numerical values such as house prices, temperature, or stock prices. Therefore, the loss function measures the **'distance' or 'difference'** between the predicted value and the actual value.

Loss Function|Concept (What does it measure?)|Features & Pros/Cons|Main Usage
---|---|---|---
nn.MSELoss|Mean squared error between prediction and actual value|Pros: Most common and standard. Penalizes large errors more by squaring them.<br>Cons: Very sensitive to outliers because of squaring. (e.g., an error of 100 becomes 10,000)|The standard loss function to try first for most regression problems.
nn.L1Loss|Mean absolute error between prediction and actual value|Pros: Uses absolute error, so less sensitive to outliers (robust).<br>Cons: Since the gradient is constant even near zero error, it can be hard to fine-tune near the optimum.|Good to use instead of MSELoss when you suspect many outliers in the data.
nn.SmoothL1Loss|Combination of MSE and L1|Pros: Behaves smoothly like MSE for small errors, and like L1 for large errors, so it's less sensitive to outliers but still stable.<br>Cons: Slightly more complex conceptually than MSE or L1.|A great alternative when you want to avoid both MSE's outlier sensitivity and L1's instability. (Often used in object detection models)


#### 2. Classification Loss Functions
Classification is a problem where you predict which category (class) the given data belongs to. Therefore, the loss function measures how different the model's predicted **'probability distribution'** is from the actual answer's probability distribution.

Loss Function|Concept (What does it measure?)|Features & Usage|Cautions
---|---|---|---
nn.CrossEntropyLoss|Difference between the model's predicted probability distribution and the actual answer's probability distribution|Feature: The standard loss function for **multi-class classification**. (e.g., classifying as dog, cat, or bird)<br>Usage: Takes logits (values before converting to probabilities) from the last layer of the model, applies Softmax internally, and computes the loss.|• Do NOT add nn.Softmax() at the end of your model! CrossEntropyLoss already includes it, so double-applying will break learning.
nn.BCELoss|Difference between the model's predicted probability and the actual answer|Feature: Loss function for **binary classification**. (e.g., pass/fail, spam/ham)<br>Usage: The model's prediction must be a probability between 0 and 1, so you must have nn.Sigmoid() at the end of your model.|• Input must be between 0 and 1. You must use nn.Sigmoid() at the end of your model.
nn.BCEWithLogitsLoss|BCELoss + Sigmoid|Feature: Combines nn.BCELoss and nn.Sigmoid into one.<br>Usage: Takes logits from the last layer of the model, applies Sigmoid internally, and computes the loss.|• Unlike BCELoss, do NOT add nn.Sigmoid() at the end of your model!<br>• More numerically stable than BCELoss, so it's recommended for binary classification.


#### Summary: Which loss function should you choose?
- Are you predicting continuous values? (Regression)
    - Yes -> If there aren't many outliers in the data, start with **nn.MSELoss**. If you're worried about outliers, use nn.SmoothL1Loss.
- Are you predicting categories? (Classification)
    - Yes -> Are there 2 categories? (Binary classification)
        - Yes -> Use nn.BCEWithLogitsLoss. (No Sigmoid at the end of the model)
    - Are there 3 or more categories? (Multi-class classification)
        - Yes -> Use nn.CrossEntropyLoss. (No Softmax at the end of the model)
