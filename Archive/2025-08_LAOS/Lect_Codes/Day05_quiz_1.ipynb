### ***1. __init__(self): The Model's "Blueprint" and "Parts Warehouse" 🏗️***
The __init__ method is a constructor that is called only once when the model object is first created. This is a space to declare and prepare in advance all the parts (materials) needed to build a model.

- Role of __init__
    - Defines and initializes the neural network layers that the model will use.
    - By analogy, it is the stage of drawing up the blueprint of a building and preparing the necessary parts such as bricks, windows, and doors in a warehouse.

- List of parts that can be prepared in __init__
    - Core Layers: Layers that perform actual calculations, such as nn.Linear, nn.Conv2d, nn.RNN, nn.LSTM, etc.
    - Activation Functions: Functions that add non-linearity to the model, such as nn.ReLU, nn.Sigmoid, nn.Softmax, etc.
    - Normalization & Dropout: Layers that stabilize learning and prevent overfitting, such as nn.BatchNorm2d, nn.Dropout, etc.
    - Pooling Layers: Layers that reduce the size of image feature maps, such as nn.MaxPool2d, nn.AvgPool2d, etc.
    - Utilities: Containers that group multiple layers into a single block, such as nn.Sequential.

### ***2. forward(self, x): The Model's "Assembly Instructions" and "Action Plan" ⚙️***
The forward method is the part that is actually executed every time the model object is called with data (x) as input. It is a space to define which parts prepared in the __init__ warehouse are brought in and in what order they are assembled to create the flow of data.

- Role of forward
    - Defines the **'calculation flow'** of how the input data is calculated from the first layer to the last layer of the model to create the final predicted value.
    - By analogy, it is like an assembly instruction manual that details how to assemble parts in order.

- List of tasks that can be performed in forward
    - Call layers declared in __init__: Use prepared parts such as x = self.layer1(x).
    - Call functional API: Functions without learnable weights, such as F.relu(x), can be called directly through the torch.nn.functional library.
    - Use basic tensor operations: All kinds of tensor manipulations are possible, such as .view(), torch.cat(), +, *.
    - Use Python control statements: You can dynamically change the calculation flow depending on the input using if and for statements.

### ***3. Layer Connection and Design: Practical Guide 💡***
Designing a model is about preparing parts in __init__ and assembling them in forward. There are two important principles at this time.

- Principle 1: Connect the pipes (match the input and output sizes)
    - This is the most important rule. The output size (out_features) of the previous layer must be the same as the input size (in_features) of the next layer.

```python
# Since layer1 outputs 5 values,
# layer2 must receive 5 values as input.
self.layer1 = nn.Linear(in_features=10, out_features=5)
self.layer2 = nn.Linear(in_features=5, out_features=1)
```

- Principle 2: Designing Hidden Layers
    The structure of the hidden layer is a key factor in determining the performance of the model and is an area of art with no right answer. However, there are some effective criteria.

    - Depth (how many to stack?):
        - Start with 1-2. Most problems are sufficient with 1-2 hidden layers.
        - You can consider stacking deeper as the problem becomes more complex.

    - Width (how many neurons?):
        - Use a funnel/pyramid structure. The most common method is to start with the input layer and gradually reduce the number of neurons (input (64) -> hidden1 (32) -> hidden2 (16) -> output (1)). This encourages the model to leave only the core of the information and compress it.
        - It is usually set to a value between the number of neurons in the input and output layers, and it is customary to use **powers of 2 (32, 64, 128...)**.

    - Activation function:
        - Use nn.ReLU() as the default. ReLU is the best activation function for hidden layers unless there is a special reason.
        - Activation functions give the model non-linearity, allowing it to solve complex problems that cannot be solved with linear layers alone.
